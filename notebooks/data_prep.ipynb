```python
# imports
import pandas as pd
import polars as pl
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import polars.selectors as cs

# sklearn imports
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import VarianceThreshold

# for saving data
import pickle
```

```python
# load data into lazyframes

#demo = pl.scan_csv("../data/raw/DemoStats_(DST)_2025_2025_GEO.csv", schema_overrides={"CODE": pl.String})
#hhs1 = pl.scan_csv("../data/raw/HouseholdSpend_(HHS)_2025_GEO_P1.csv", schema_overrides={"CODE": pl.String})
#hhs2 = pl.scan_csv("../data/raw/HouseholdSpend_(HHS)_2025_GEO_P2.csv", schema_overrides={"CODE": pl.String})


demo = pl.scan_csv("../data/DemoStats_2025_CSV_fixed/DemoStats_(DST)_2025_2025_GEO.csv", schema_overrides={"CODE": pl.String})
hhs1 = pl.scan_csv("../data/HouseholdSpend_2025_CSV_fixed/HouseholdSpend_(HHS)_2025_GEO_P1.csv", schema_overrides={"CODE": pl.String})
hhs2 = pl.scan_csv("../data/HouseholdSpend_2025_CSV_fixed/HouseholdSpend_(HHS)_2025_GEO_P2.csv", schema_overrides={"CODE": pl.String})

# merge hhs1 and hhs2
hhs = hhs1.join(hhs2, on=["CODE", "GEO"], how="inner")
```

```python
# show demo

demo.head().collect()
```

```python
# filter demo to only include rows where GEO contains "PRCDDA"

demo = demo.filter(pl.col("GEO").str.contains("PRCDDA"))

demo.head().collect()
```

```python
# show hhs

hhs.head().collect()
```

```python
# filter hhs to only include rows where GEO contains "PRCDDA"

hhs = hhs.filter(pl.col("GEO").str.contains("PRCDDA"))

hhs.head().collect()
```

```python
# define HBR (HSSH001S/HSAGDISPIN)

hhs = hhs.with_columns((pl.col("HSSH001S") / pl.col("HSAGDISPIN")).alias("HBR"))

# if HSAGDISPIN is 0, set HBR to None (bc it would be inf)
hhs = hhs.with_columns(
    pl.when(pl.col("HSAGDISPIN") == 0).then(None).otherwise(pl.col("HBR")).alias("HBR")
)

# drop rows with none values (HBR)
hhs = hhs.filter(pl.col("HBR").is_not_null())

hhs.head().collect()
```

```python
# remove columns with HSSH

hhs = hhs.drop(cs.contains("HSSH"))

hhs.head().collect()
```

```python
# remove aggregates

hhs = hhs.drop(pl.col("HSAGDISPIN"))
hhs = hhs.drop(pl.col("HSAGDISCIN"))
hhs = hhs.drop(pl.col("HSHNIAGG"))
hhs = hhs.drop(pl.col("HSTT001"))
hhs = hhs.drop(pl.col("HSTE001"))
hhs = hhs.drop(pl.col("HSTX001"))
hhs = hhs.drop(pl.col("HSTC001"))

hhs.head().collect()
```

```python
# merge demo and hhs on CODE and GEO

data = hhs.join(demo, on="CODE", how="inner")

data.head().collect()
```

```python
# remove labels

data = data.drop(pl.col("CODE"))
data = data.drop(pl.col("GEO"))
data = data.drop(pl.col("GEO_right"))

data = data.collect()

data.head()
```

```python
# remove 0 variance col (done before splitting bc its the same between both sets anyways)

selector = VarianceThreshold(threshold=0)
selector.fit(data)
selected_cols = [col for col, keep in zip(data.columns, selector.get_support()) if keep]
data = data.select(selected_cols)

data.head()
```

```python
# split into train and test

X_train, X_test, y_train, y_test = train_test_split(data.drop('HBR'),
                                                    data.select('HBR'),
                                                    test_size = 0.2,
                                                    random_state = 42)

y_train = y_train.to_pandas().squeeze()
y_test = y_test.to_pandas().squeeze()

display(X_train.head())
display(y_train.head())
```

```python
# handle missing values (calculate on train apply to test)

imputer = SimpleImputer(strategy='median')
X_train = pd.DataFrame(imputer.fit_transform(X_train), columns=X_train.columns)
X_test = pd.DataFrame(imputer.transform(X_test), columns=X_test.columns)
```

```python
# fit scaler on train apply to test

scaler = StandardScaler()
X_train_scaled = pd.DataFrame(scaler.fit_transform(X_train), columns=X_train.columns)
X_test_scaled = pd.DataFrame(scaler.transform(X_test), columns=X_test.columns)
```

```python
display(X_train.head())
display(y_train.head())
```

```python
# save data

with open('../data/processed/tableau.pkl', 'wb') as f:
    pickle.dump({
        'X_train': X_train,
        'X_test': X_test,
        'X_train_scaled': X_train_scaled,
        'X_test_scaled': X_test_scaled,
        'y_train': y_train,
        'y_test': y_test,
        'scaler': scaler,
        'imputer': imputer
    }, f)
```
